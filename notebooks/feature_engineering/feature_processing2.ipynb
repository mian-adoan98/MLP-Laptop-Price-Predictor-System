## Data Transformation

```python
# Extract ram from dataset
ram_feature_df = dataset[["Ram Size"]]
ram_feature_df.loc[:, "Ram Size"] = ram_feature_df["Ram Size"].fillna(0.0)
ram_feature_df.loc[:, "Ram Size"] = ram_feature_df["Ram Size"].replace(0.0, round(ram_feature_df["Ram Size"].mean(), 2))
ram_feature_df.loc[:, "Ram Size"] = ram_feature_df["Ram Size"].apply(lambda x: round(x, 2))
ram_feature_df
```
